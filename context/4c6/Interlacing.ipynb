```python
import pydvma as dvma
import numpy as np
import matplotlib
import matplotlib.pyplot as plt
matplotlib.rcParams.update({'font.size': 12,'font.family':'serif'})
```

```python
d = dvma.DataSet()
```

```python
N1 = 3
N2 = 2

# randomise amplitudes in range -1 to +1
an1 = np.random.uniform(-1,1,N1)
an2 = np.random.uniform(-1,1,N2)

# randomise frequencies in range 0-10Hz
wn1 = np.random.rand(N1) * 10
wn2 = np.random.rand(N2) * 10

# frequency axis to view
w = np.linspace(0,10,1000)
```

```python
G1 = 0
for n in range(N1):
    G1 += 1 / (wn1[n]**2 - w**2)
    
G2 = 0
for n in range(N2):
    G2 += 1 / (wn2[n]**2 - w**2)

GT = 1/(1/G1 + 1/G2)
```

```python
fig,ax = plt.subplots(1,1,figsize = (9,6))
ax.plot(w,G1)
ax.plot(w,G2)
ax.plot(w,GT)
ax.set_ylim(-0.2,0.2)
```
