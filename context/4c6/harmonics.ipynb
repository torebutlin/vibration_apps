```python
import numpy as np
from IPython.display import Audio

import matplotlib.pyplot as plt

# Time parameters
fs = 44100  # Sampling frequency
T = 3
t = np.arange(0, T, 1/fs)

# Four sinusoids with different frequencies and decay rates
freq1 = 50
freq_ratios = [1,2,3]
freq = freq1 * np.array(freq_ratios)
print(freq)
time_const = 2  # Decay time constant

# Create three audio signals
# Signal 1: x1 freq only
signal1 = np.zeros_like(t)
decay = np.exp(-t / time_const)
signal1 = decay * np.sin(2 * np.pi * freq[0] * t)
signal1 = signal1 / np.max(np.abs(signal1))  # Normalize

# Signal 2: x2 freq only
signal2 = np.zeros_like(t)
signal2 = decay * np.sin(2 * np.pi * freq[1] * t)
signal2 = signal2 / np.max(np.abs(signal2))  # Normalize

# Signal 3: x2 and x3 combined
signal3 = np.zeros_like(t)
signal3 += decay * np.sin(2 * np.pi * freq[1] * t)
signal3 += decay * np.sin(2 * np.pi * freq[2] * t)
signal3 = signal3 / np.max(np.abs(signal3))  # Normalize

# Convert signals to audio format
audio_signal1 = np.int16(signal1 / np.max(np.abs(signal1)) * 32767)
audio_signal2 = np.int16(signal2 / np.max(np.abs(signal2)) * 32767)
audio_signal3 = np.int16(signal3 / np.max(np.abs(signal3)) * 32767)

# Play the sounds
print(f"Signal 1 ({freq[0]} Hz):")
display(Audio(audio_signal1, rate=fs))
print(f"Signal 2 ({freq[1]} Hz):")
display(Audio(audio_signal2, rate=fs))
print(f"Signal 3 ({freq[1]} Hz and {freq[2]} Hz):")
display(Audio(audio_signal3, rate=fs))
```
