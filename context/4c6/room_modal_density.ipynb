```python
import numpy as np
import IPython.display as ipd
```

```python
# Define parameters
f = 500  # frequency in Hz
L1 = 6  # length in meters
L2 = 4  # width in meters
L3 = 2.5  # height in meters
V = L1 * L2 * L3  # volume in cubic meters
S = 2 * (L1 * L2 + L2 * L3 + L1 * L3)  # surface area in square meters
L = 4 * (L1 + L2 + L3)  # perimeter length in meters
c = 340

# Calculate N(f) based on the formula:
# N(f) = (4πV/3c³)f³ + (πS/4c²)f² + (L/8c)f

term1 = (4 * np.pi * V / (3 * c**3)) * f**3
term2 = (np.pi * S / (4 * c**2)) * f**2
term3 = (L / (8 * c)) * f

N_value = term1 + term2 + term3
print(f"term1: {term1:.3g}")
print(f"term2: {term2:.3g}")
print(f"term3: {term3:.3g}")
print("-------------------")
print(f"N(f): {N_value:.3g}")

# Generate tone 
duration = 4  # seconds
sample_rate = 44100
t = np.linspace(0, duration, int(sample_rate * duration))
tone = np.sin(2 * np.pi * f * t)

# Apply smooth start/stop window (Hann window)
window = np.hanning(len(tone))
tone = tone * window

# Display audio
ipd.Audio(tone, rate=sample_rate)
```

```python
# Calculate Schroeder frequency
f_schroeder = 2000 * np.sqrt(0.6 / V)
print("Schroeder frequency:", f"{f_schroeder:.3g}", "Hz")

N_modes_at_schroeder = (4 * np.pi * V / (3 * c**3)) * f_schroeder**3 + \
                       (np.pi * S / (4 * c**2)) * f_schroeder**2 + \
                       (L / (8 * c)) * f_schroeder
print("N(f) at Schroeder frequency:", int(round(N_modes_at_schroeder)))
```

```python

# Generate tone at Schroeder frequency
duration = 4  # seconds
sample_rate = 44100
t = np.linspace(0, duration, int(sample_rate * duration))
tone = np.sin(2 * np.pi * f_schroeder * t)

# Apply smooth start/stop window (Hann window)
window = np.hanning(len(tone))
tone = tone * window

# Display audio
ipd.Audio(tone, rate=sample_rate)
```
