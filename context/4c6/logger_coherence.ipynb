```python
import pydvma as ma
import matplotlib.pyplot as plt
```

```python
%matplotlib widget
```

```python
settings = ma.MySettings(channels=1,
                         fs=44100,
                         stored_time=15,
                         pretrig_samples=200,
                         viewed_time=3,
                         chunk_size=1024,
                         device_driver='soundcard',
                         device_index=2)
logger = ma.Logger(settings)
```

```python
osc = ma.Oscilloscope(settings)
```

```python
m = logger.dataset.modal_data_list[0]
fn = m.fn
Q = 1/2/m.zn
plt.plot(fn,Q)
plt.xlabel('f_n (Hz)')
plt.ylabel('Q')
```
