# Bessel functions

```python
import numpy as np
import scipy.special as special
import matplotlib
import matplotlib.pyplot as plt
matplotlib.rcParams.update({'font.size': 12,'font.family':'serif'})
```

```python
%matplotlib notebook
```

```python
N = 10
kr = np.linspace(0,20,1000)
u = [] # make an empty list
for n in range(N):
    u += [special.jv(n,kr)] # add bessel functions to the list
```

```python
# plot
fig,ax = plt.subplots(figsize=(9,6))
for n in range(N):
    ax.plot(kr,u[n],label='n = ' + str(n))

ax.set_xlabel('$kr$')
ax.set_ylabel('$J_n(kr)$')
ax.grid(True)
ax.legend()
ax.set_title('Bessel functions of the first kind')
```

```python
# plot
fig,ax = plt.subplots(2,2,figsize=(9,6),sharex=True,sharey=True)
for n in range(4):
    axn = ax.reshape(-1,1)[n,0]
    axn.plot(kr,u[n],label='n = ' + str(n))
    axn.set_title(f'n = {n}')
    axn.grid(True,which='major')
    axn.set_xlim([0,10])
    axn.set_ylim([-0.5,1])
    axn.set_yticks([-0.5,0,0.5,1])
    
    zeros = special.jn_zeros(n,7)
    for z in zeros:
        if z<10:
            if n<3:
                offset = 0.2
            else:
                offset = 1
            axn.text(z-offset,0.62,f'{z:.2f}',fontsize=12)
            axn.plot(z,0,'xk')
            axn.plot([z,z],[0,0.6],'--k',linewidth=1)

fig.text(0.5, 0.04, '$kr$', ha='center')
fig.text(0.04, 0.5, '$J_n(kr)$', va='center', rotation='vertical')

```
