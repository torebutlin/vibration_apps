# First order damped system analysis

```python
import numpy as np
from scipy import linalg as la
import matplotlib
import matplotlib.pyplot as plt
matplotlib.rcParams.update({'font.size': 12,'font.family':'serif'})
```

```python
%matplotlib notebook
```

## Parameter choices

```python
m = 1
k = 1
c = 0.1
```

## Construct M, C, K

```python
M = m*np.eye(3)

C = np.zeros((3,3))
C[0,0]=c
C[1,1] = c
C[0,1] = -c
C[1,0] = -c


K = 2*k*np.eye(3)
K += -k*np.diag(np.array([1,1]),1)
K += -k*np.diag(np.array([1,1]),-1)

print('M = ')
print(M)

print('\nC = ')
print(C)

print('\nK = ')
print(K)
```

## undamped case - second order

```python
wn2,Phi = la.eigh(K,M)  # eigh tells scipy that the matrices are symmetric
wn = np.sqrt(wn2)       # sqrt to get w from w**2
scale_factor = np.max(np.abs(Phi),axis=0) # get max of each column
Phi = Phi/scale_factor  # normalise so numbers easier to interpret

print('wn = ')
print(np.round(wn*100)/100)  # round(*100)/100 to print 2 decimal places

print('\n Phi = ')
print(np.round(Phi*100)/100) # round(*100)/100 to print 2 decimal places
```

## undamped case - first order method

```python
A_undamped = np.block([[np.zeros((3,3)),np.eye(3)],[-np.matmul(la.inv(M),K),-np.matmul(la.inv(M),C)*0]])
print('A_undamped = ')
print(A_undamped)
```

```python
lambda_A_undamped, Phi_A_undamped = la.eig(A_undamped)
print('Complex lambda from the undamped A matrix = ')
print(np.round(lambda_A_undamped*100)/100)

print('\nImaginary part of complex lambda from the undamped A matrix, wn = ')
wn_A_undamped = np.imag(lambda_A_undamped)
print(np.round(wn_A_undamped*100)/100)

print('\nMode shapes Phi from A (scaled to max of first three rows in each column) = ')
scale_factor = np.max(np.abs(Phi_A_undamped[0:3,:]),axis=0) # get max of each column from first three rows
Phi_A_undamped = Phi_A_undamped/scale_factor  # normalise so numbers easier to interpret
print(np.round(Phi_A_undamped*100)/100)
```

## damped case - first order method

```python
A = np.block([[np.zeros((3,3)),np.eye(3)],[-np.matmul(la.inv(M),K),-np.matmul(la.inv(M),C)]])
print('A = ')
print(A)
```

```python
lambda_A, Phi_A = la.eig(A)
print('Complex lambda from the **damped** A matrix = ')
print(np.round(lambda_A*100)/100)

print('\nImaginary part of complex lambda from the **damped** A matrix, wn = ')
wn_A = np.imag(lambda_A)
print(np.round(wn_A*100)/100)

print('\nReal part of complex lambda from the **damped** A matrix, sigma = ')
sigma_A = np.real(lambda_A)
print(np.round(sigma_A*1000)/1000)

print('\nComplex mode shapes Phi from the **damped** A matrix (scaled) = ')
scale_factor = np.max(np.abs(Phi_A[0:3,:]),axis=0) # get max of each column from first three rows
Phi_A = Phi_A/scale_factor  # normalise so numbers easier to interpret
print(np.round(Phi_A*100)/100)
```
