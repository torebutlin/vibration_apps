```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
w = np.linspace(0, 10, 100000)
an = [1]
zn =[0.1]
wn = [1.25]
for i in range(len(an)):
    if i == 0:
        H = 1j*w*an[i]/(wn[i]**2-w**2 + 2j*zn[i]*wn[i]*w)
    else:
        H += 1j*w*an[i]/(wn[i]**2-w**2 + 2j*zn[i]*wn[i]*w)

G = 1 / H

f1 = wn[0] - wn[0]*zn[0]
f2 = wn[0] + wn[0]*zn[0]

plt.figure(figsize=(10, 8), dpi=150)
plt.plot(np.real(H), np.imag(H))
plt.plot(np.real(1j*w*H), np.imag(1j*w*H))
plt.plot(np.real(H/1j/w), np.imag(H/1j/w))
# Add crosses at specific frequency points with labels
freq_points = [1.23,1.24,1.25,1.26,1.27,f1,f2]
for freq in freq_points:
    idx = np.argmin(np.abs(w - freq))
    
    # Plot crosses
    plt.plot(np.real(H[idx]), np.imag(H[idx]), 'kx', markersize=8)
    plt.plot(np.real(1j*w[idx]*H[idx]), np.imag(1j*w[idx]*H[idx]), 'kx', markersize=8)
    plt.plot(np.real(H[idx]/1j/w[idx]), np.imag(H[idx]/1j/w[idx]), 'kx', markersize=8)
    
    # Add text label only for H plot
    H_point = (np.real(H[idx]), np.imag(H[idx]))
    angle = np.arctan2(H_point[1], H_point[0])
    offset_distance = 10
    x_offset = offset_distance * np.cos(angle)
    y_offset = offset_distance * np.sin(angle)
    
    plt.annotate(f'{freq}', H_point, xytext=(x_offset, y_offset), 
                textcoords='offset points', fontsize=8)

plt.gca().set_aspect('equal', adjustable='box')
plt.grid(True)
plt.xlabel('Real')
plt.ylabel('Imaginary')
plt.xlim(-3, 5)

1/4/0.1/1.25
```
