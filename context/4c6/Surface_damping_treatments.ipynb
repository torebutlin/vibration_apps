# Surface damping treatments

```python
import numpy as np
import matplotlib
import matplotlib.pyplot as plt
matplotlib.rcParams.update({'font.size': 12,'font.family':'serif'})
```

```python
%matplotlib notebook
```

## steel properties

```python
L = 1         # length
h1 = 0.005    # thickness

E1 = 210e9    # Young's modulus
G1 = 3/8*E1   # Shear modulus
I1 = h1**3/12 # Second moment of area per unit width
```

## damping layer properties

```python
h2 = np.linspace(1e-6,5e-3,100)   # thickness range to investigate
loss = 1                      # loss factor
E2 = 1e9  *(1+loss*1j)          # complex young's modulus
G2 = 1e9/3 *(1+loss*1j)          # complex shear modulus
```

## constrained layer properties

```python
h3 = 0.0005      # thickness

E3 = np.copy(E1) # Young's modulus
G3 = np.copy(G1) # Shear modulus
```

## Two-layer formula

```python
h = h2/h1
e = E2/E1

EI_2_layer = E1*I1*(1 + e*h**3 + 3*(1+h)**2*e*h/(1+e*h))

eta_2_layer = np.imag(EI_2_layer) / np.real(EI_2_layer)
```

## Three-layer formula

```python
n = 5 # MODE NUMBER
def calc_three_layer(n):
    p = n*np.pi / L
    h21 = (h1+h2)/2
    h31 = (h1+h3)/2 + h2
    g = G2/(E3*h3*h2*p**2)

    d = (E2*h2*(h21-h31/2) + g*(E2*h2*h21 + E3*h3*h31))/(E1*h1+E2*h2/2 + g*(E1*h1+E2*h2+E3*h3))

    EI_3_layer = E1*h1**3/12 + E2*h2**3/12 + E3*h3**3/12
    EI_3_layer += -E2*h2**2/12*(h31-d)/(1+g) + E1*h1*d**2
    EI_3_layer += E2*h2*(h21-d)**2 + E3*h3*(h31-d)**2
    EI_3_layer += -(E2*h2/2*(h21-d) + E3*h3*(h31-d)) * (h31-d)/(1+g)

    eta_3_layer = np.imag(EI_3_layer) / np.real(EI_3_layer)
    return eta_3_layer
```

```python
fig,ax = plt.subplots(figsize=(9,6))
ax.plot(h2*1e3,eta_2_layer,'k--',label='free layer')
for n in range(1,6):
    eta_3_layer = calc_three_layer(n)
    ax.plot(h2*1e3,eta_3_layer,label=f'constrained layer, n = {n}')
ax.set_xlabel('Thickness of damping layer (mm)')
ax.set_ylabel('Effective $\eta$')
ax.legend()
ax.set_title('Effective loss factors for free- and constrained-layer damping treatments')
ax.grid(True)
ax.autoscale(enable=True, axis='x', tight=True)
```
