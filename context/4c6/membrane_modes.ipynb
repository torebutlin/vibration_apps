# Rectangular membrane modes

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib
matplotlib.rcParams.update({'font.size': 12,'font.family':'serif'})
```

```python
N = 1000
Lx = 1
Ly = 1
x = np.linspace(0,Lx,N)
y = np.linspace(0,Ly,N)
X,Y = np.meshgrid(x,y)
pi = np.pi
```

```python
nx = 3
ny = 2
U = np.sin(nx*pi*X/Lx) * np.sin(ny*pi*Y/Ly)
```

```python
plt.figure()
plt.pcolormesh(X,Y,U,cmap='viridis')
```

```python
from mpl_toolkits.mplot3d import Axes3D
from matplotlib.animation import FuncAnimation
from IPython.display import HTML

fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

def update(frame):
    ax.clear()
    U = np.sin(nx * pi * X / Lx) * np.sin(ny * pi * Y / Ly) * np.cos(frame * pi / 25)
    ax.plot_surface(X, Y, U, cmap='gray', shade=True)
    ax.set_zlim(-1, 1)
    return ax,

ani = FuncAnimation(fig, update, frames=50, interval=20, blit=False)
HTML(ani.to_html5_video())
```
