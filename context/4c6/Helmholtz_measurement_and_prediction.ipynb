# Prediction

```python
# without end correction
c = 340

r1 = 0.02 / 2
L1 = 0.08

r2 = 0.065/2
L2 = 0.22

S = np.pi*r1**2
V = np.pi*r2**2*L2

w = c*np.sqrt(S/V/L1)
print(w/2/np.pi)
```

```python
# with end correction
L1 = 0.08+(0.6+0.85)*r1 # flange and 
w = c*np.sqrt(S/V/L1)
print(w/2/np.pi)
```
